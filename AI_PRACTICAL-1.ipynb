```python
print("Hello! I am cleaning robo.")
print("Today we will build a cleaning robot together:")
print(2 + 3, "is the answer to 2 + 3")

room = ["D", "C", "D", "D", "C"]

def show_room(room):
    picture = ""
    for spot in room:
        if spot == "D":
            picture += "💩 "
        else:
            picture += "✨ "
    print(picture)

print("Our room right now:")
show_room(room)

def clean_spot(spot):
    if spot == "D":
        return "C"
    else:
        return "C"

result = clean_spot("D")
print("The robot looked at a dirty spot 💩 and made it:", result,
      "(C means clean ✨)")

print("BEFORE - the dirty room:")
show_room(room)

for i in range(len(room)):
    room[i] = clean_spot(room[i])
    print("After cleaning spot number", i + 1, ":")
    show_room(room)

print()
print("AFTER - all done! ⭐⭐⭐⭐⭐⭐")
```

```python
print("🤖 We are creating a ChatBot AI!")

while True:
    user = input("You: ").lower()

    if user == "hi" or user == "hello":
        print("🤖 ChatBot: Hello! How are you?")

    elif user == "i am fine" or user == "iam fine":
        print("🤖 ChatBot: That's good! What is today's agenda?")

    elif "python" in user:
        print("🤖 ChatBot: That's great! Python is a high-level, easy-to-read programming language designed to be simple and powerful.")

    elif "what is your name" in user:
        print("🤖 ChatBot: My name is Python ChatBot.")

    elif user == "bye":
        print("🤖 ChatBot: Bye! Have a nice day!")
        break

    else:
        print("🤖 ChatBot: Sorry, I don't understand.")
```

```python
# List of spam keywords
spam_words = [
    "lottery",
    "winner",
    "prize",
    "free",
    "offer",
    "money",
    "click",
    "claim"
]

# Get email from user
email = input("Enter your email message:\n")

# Convert email to lowercase
email = email.lower()

# Check for spam words
spam = False

for word in spam_words:
    if word in email:
        spam = True
        break

# Display result
if spam:
    print("\nResult: Spam Mail")
else:
    print("\nResult: Not Spam Mail")


```
