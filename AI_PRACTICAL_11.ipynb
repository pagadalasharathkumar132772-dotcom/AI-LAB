```python

# Logical Inference using Propositional Logic

fact = {"raining"}

print("Initial fact:", fact)

rules = {
    "raining": "Road is wet",
    "Road is wet": "drive carefully",
    "drive carefully": "speed reduce"
}

# Apply inference rules
changed = True

while changed:
    changed = False

    for condition, conclusion in rules.items():
        if condition in fact and conclusion not in fact:
            fact.add(conclusion)
            print("Derived:", conclusion)
            changed = True

print("\nFinal facts:", fact)
```
